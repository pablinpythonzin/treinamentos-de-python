```python
#Usando o metodo Pop
linguagem = ["PHP","SQL","Python"]

valor_retirado = linguagem.pop()#Podemos definir o indice que queremos pegar tambem e podemos deixar em uma variavel esse idnice pego

print(linguagem)
print(valor_retirado) #Sendo assim podemos ver o valor retirado por meio da variavel, seja ela o metodo reitrado por indice ou sem o indice
```

```python
lista_salgado = ["Coxinha","cachorro-quente","mandioca","bolo de chocolate"] #lista simpls normal

intruso = "bolo de chocolate" #criamos uma variavel "intruso" na lista de salgado

lista_salgado.remove(intruso)#aqui pegamos a lista de salgado primeiro, usamos o metodo remove e dentro do remove colocaos a variavel intruso e assim a nossa lista de salgado fica correta sem o doce
print(lista_salgado)
```

```python
#Usando o Metodo Sort

lista_nomes = ["pablo","ana","miguel"]
lista_nomes.sort()#o metodo sort deixa em ordem alfabetica permanente, se todos estiverem com letras minusculas
print(lista_nomes)
```

```python
#usando a ordem alfabetica reversa com argumentojunto com o metodo sort

lista_carros = ["bmw","audi","toyota","honda"]
lista_carros.sort(reverse=True) #Argumento passando dentro do metdo indicando o alfabeto reverso, se caso usar maiuscula em uma das letras, sai da ordem alfabetica
print(lista_carros)#A lista fica permanentemente nessa ordem
```

```python
#Usando a função Sorted
lista_nomes = ["pablo","ana","miguel","mateus"]

print(sorted(lista_nomes))#Aqui usamos uma função aonde ela não mexe direto na lista mais apenas muda a forma do print

print(lista_nomes)#ainda que usamos a função ela não alterou nada na lista original

#resumo a função diferente de metodo é como se a função apenas formatesse na hora do print enquanto o metodo ele pode mudar listas permanentemente ou em variaveis
```

```python
#usando o metodo reverse

comida = ["arroz","feijão","carne"]

comida.reverse() #Revertendo a ordem da lista comida
print(comida)

```

```python
#Usando a função Len

senha = "carro"
print(len(senha)) #essa função ela serve para medir o tamanho da palavra, ela começa a contar pelo 1 e assim adiante
```
