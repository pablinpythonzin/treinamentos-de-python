```python
# TUPLA É AGLO IMUTÁVEL

dimensions = (200 , 50)
print(dimensions[1])
print(dimensions[0])

dimensions[0] = 12 #AQUI OCORRE UMA TENTATIVA DE MUDAR AS DIMENSÕES DO QUADRADO
print(dimensions) #NÃO OCORRE ESSA MUDANÇA POR CAUSA DA AÇÃO DA TUPLA

#NOTA: AS TUPLAS SÃO DEFINIDAS PELA PRESENÇA DE UMA VIRGULA
```

```python
# TUPLA COM LAÇO FOR
print("Original")
dimension = (100 , 200)
for dimensions in dimension:
  print(dimensions)

print("\nmodificado")
dimension = (200 , 300)
for dimensions in dimension:
  print(dimensions)
```
