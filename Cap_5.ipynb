```python
current_users = ["Toji", "Gojo", "Itadori", "Sukuna", "Negumi"]
new_users = ["Deku", "Shigaraki", "Dabi", "Toji", "Gojo"]

# Dica do livro: Criamos uma lista com os usuários atuais todos em letras minúsculas.
# Isso garante que se alguém tentar cadastrar "toji" ou "TOJI", o sistema ainda detecte como repetido.
current_users_lower = [user.lower() for user in current_users]

for verificar_repetido in new_users:
    # Convertemos o nome novo para minúsculo antes de comparar
    if verificar_repetido.lower() in current_users_lower:
        # Usamos f-string para mostrar exatamente QUAL nome já está sendo usado
        print(f"O nome '{verificar_repetido}' já está sendo usado. Você precisará escolher outro nome.")
    else:
        print(f"O nome '{verificar_repetido}' está disponível. Tudo ok !!!")
```

```python
lista_preco = [500,1500,2000,100,25]

nova_lista_precos = []

for preco in lista_preco:
  nova_lista_precos.append(preco * 2)
print(nova_lista_precos)
```

```python
nova_lista_precos1 = [preco * 2 for preco in lista_preco]

print(nova_lista_precos1)
```

```python
lista_preco = [500,1500,2000,100,25]

produtos_impostos = []

for preco in lista_preco:
  if preco >= 1000:
    produtos_impostos.append(preco * 0.5)
    print(f"Pague imposto {preco}\nA lista dos produtos com imposto é {produtos_impostos}")
```

```python
 lista_preco = [500,1500,2000,100,25]

produtos_impostos = [preco * 0.5  for preco in lista_preco if preco > 1000]

print(produtos_impostos)
```
