```python
#Retornando um valor simples
#usando Return

def pegando_nome_inteiro(primeiro_nome , segundo_nome):
  nome_inteiro = f"{primeiro_nome} {segundo_nome}" #a função combina os dois nomes para uma variavel
  return nome_inteiro.title()#coverte em title e é retornado a linha de chamada

nome = pegando_nome_inteiro('pablo' , 'mota')#quando chamamos uma função que retorna um valor, é necessario fornecer uma variavel à qual o valor de retorno possa ser atribuído, para assim darmos outros nomes de variaveis usando o retorno da função
print(nome)
```

```python
#definindo um argumento como opcional

def pegando_nome_inteiro(primeiro_nome , nome_do_meio , ultimo_nome):
  nome_inteiro = f"{primeiro_nome} {nome_do_meio} {ultimo_nome}"
  return nome_inteiro.upper()

nome = pegando_nome_inteiro('pablo' , 'mota' , 'oliveira')
print(nome)
```

```python
#Usando condição na função para adicionar nome do meio como opcional

def pegando_nome_inteiro(primeiro_nome , ultimo_nome , nome_do_meio=''):#no parametro "nome_do_meio" se tiver algo preenchido como string é passando como True
  if nome_do_meio: #aqui ele faz uma comparação se caso tiver argumento no parametro opcional
    nome_inteiro = f"{primeiro_nome} {nome_do_meio} {ultimo_nome}"#aqui pode se tornar 'false' e cair para else se caso o parâmetro "nome_do_meio" não for preenchido
  else:
    nome_inteiro = f"{primeiro_nome} {ultimo_nome}"
  return nome_inteiro.upper()#como o return esta fora dos blocos "if" e "else" então ele vai retornar o valor de acordo com a condição, seja com o nome inteiro ou sem o nome do meio


nome = pegando_nome_inteiro('pablo' , 'mota' , 'a')
print(nome)
```

```python
#Retornando um dicionário

def construindo_pessoa(first_name , last_name):
  pessoa = {
      'first': first_name ,
      'last': last_name,
  }

  return pessoa

pessoa_1 = construindo_pessoa('pablo' , 'mota')
pessoa_2 = construindo_pessoa('julia' , 'lacerda')

print(pessoa_1 , pessoa_2)
```

```python
#adicionando parâmetro de idade

def construindo_pessoa(first_name , last_name , age=None):
    pessoa = {'primeiro_nome': first_name , 'ultimo_nome': last_name,}

    if age: #aqui ele vai verificar se o parâmetro 'age' vai receber algum valor no argumento
      pessoa['idade'] = age #se caso tiver ele criará uma chave 'idade'
    return pessoa



pessoa1 = construindo_pessoa('pablo' , 'mota' , ) #não é obrigado a usar o parâmetro 'age' por estar como 'vazio'
pessoa2 = construindo_pessoa('miguel' , 'oliveira' , age=17)#adicionando o parâmetro do 'age' como opcional por ser opcional

print(pessoa1 , pessoa2)
```

```python
#Usando um função com um loop While

def construindo_pessoa(primeiro_nome , ultimo_nome):
  nome_inteiro = f"{primeiro_nome} {ultimo_nome}"
  return nome_inteiro.upper()

while True:
  print("Escreva 'sair' para parar")

  p_nome = input("digite seu primeiro nome: ")
  if p_nome == 'sair':
    break

  u_nome = input("digite seu ultimo nome: ")
  if u_nome == 'sair':
    break
  

  nome_formatado = construindo_pessoa(p_nome , u_nome)
  print(nome_formatado)
```
