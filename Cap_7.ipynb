```python
#Usando FLAGS

#Em Python, uma variável flag é uma variável booleana (True ou False) usada para indicar se uma condição foi atendida ou não durante a execução do programa.

prompt = "Retorne para mim quando puder "
prompt += "escreva 'quit' para sair: "

ativo = True #a variavel esta sendo "True" então esta ativa
while ativo:
  msg = input(prompt)#sua resposta vai estar na variavel "msg"

  if msg == "quit":
    ativo = False
    #se caso a palavra for "quit" o codigo vai acabar e a variavel "ativo" será modificada
  else:
    print(msg) #vai mostrar a sua ultima respota enquanto voce tiver errando
```

```python
#Usando a opção BREAK
#BREAK serve para encerrar o loop

prompt = "\nDigite um nome o nome da cidade que você já visitou:"
prompt += "\nAperte 'quit' para finalizar.\n"

while True:
  city = input(prompt)

  if city == 'quit':
    break

  else:
    print(f"eu amo ir para {city.title()}")
```

```python
#Usando o CONTINUE em loop
#Serve para continuar a rodar o codigo independente se uma condição ja foi rodada

numero = 0 #numero inicial
while numero <10:
  numero += 1

  if numero % 2 == 0: #breve verificação se caso o resto da divisão for 0
    continue #ele vai ignorar os numeros 2 4 6 8 10 por terem resto 0

  print(numero)
  #com esse continue vai ignorar os nuemros par, então quando o contador começar no 1 vai ser validado depois subiu pro 2 vai ser pulando por terminar com 0 e assim ele vai continuar e não parar o codigo
```

```python
usuarios_indefinidos = ['gojo',"toji",'deku']
usuarios_confirmados = []

while usuarios_indefinidos: #aqui o loop vai rodar até a lista dos usuarios indefinidos ficar vazia
  usuarios_atuais = usuarios_indefinidos.pop() #a lista dos usuarios indefinidos vai ficar em uma variavel

  print(f"verificando usuarios: {usuarios_atuais.title()}")
  usuarios_confirmados.append(usuarios_atuais)#adicionando os usuarios indefinios na lista dos usuarios confirmados

print("Usuarios confirmados")
for usuario_confirmado in usuarios_confirmados: #mostra cada usuario na lista dos usuarios confirmados
  print(usuario_confirmado.title())
```

```python
# Usando metodo REMOVE() com WHILE

pets = ['dog','cat','goldfish','cat','dog']
print(pets)

while 'cat' in pets: #enquanto "cat" tiver na lista de "pets"
  pets.remove("cat") #removendo "cat" na lista
print(pets)
```

```python
#Usando WHILE junto com Dicionario

respostas = {}
#definindo uma flag
fluxo = True

while fluxo: #enquanto tiver rodando
  nome = input("Qual é seu nome: ")
  resposta = input("Qual montanha voce subiria: ")

  #armazenando chave e valor de acordo com o usuario
  respostas[nome] = resposta

  #verificação se caso algume vai participar
  repetir = input("Alguem vai participar da pergunta? (sim/ não) ")
  if repetir == 'não':
    fluxo = False

#Pesquisa completa
print("Resultados")
for nome , resposta in respostas.items(): #definindo chave e valor no dicionario
  print(f"{nome} você gostaria de subir em qual montanha {resposta}")
```

```python
# 7.8 aprendendo a usar o while corretamente

livros_devolvidos = ['python', 'algoritmos', 'clean Code', 'redes']
livros_na_estante= []

while livros_devolvidos:
  livro = livros_devolvidos.pop()
  print(f"livro {livro} sendo devolvido")

  livros_na_estante.append(livro)

  print(f"Foram devolvidos {len(livros_na_estante)} livros que são {livros_na_estante}")

for livros in livros_na_estante:
  print(f"os livros devolvidos {livros} ")


```
