```python
#alterando valores com if-elif-else

# usando uma verificação de if-elif-else no valor da chave "speed"

# "X" é o valor inicial fica embaixo, ele ira mover pra cima ou pra baixo
# "Y" é o valor que irá mover para direita ou esquerda

alien = {'x_posição': 0 , #valor inicial
         'y_posição': 25 , #movimetou para a direita
        'speed': 'medium' #velocidade
         }

print(f"posição inicial do alien {alien['x_posição']}")

#Alterando a chave "SPEED"
alien['speed'] = "fast" #alterando o valor da chave do dicionario

#valores da chave "speed" que passaram no teste
if alien['speed'] == "slow":
  x_implementar = 1

elif alien['speed'] == "medium":
  x_implementar = 2

elif alien['speed'] == "fast": #conforme o valor da chave é "fast" o teste passa
  x_implementar = 3

#a posisão inicial do alen vai ser: a posição x do alien mais a condição de speed
alien['x_posição'] = alien['x_posição'] + x_implementar

print(f"A nova posição do Alien foi para {alien['x_posição']}")
```

```python
favorite_languages = {
    'jen':'python',
    'sarah':'c',
    'edward': 'rust',
    'phil': 'python',
}

friends = ['jen','phil']

#nessa parte ele vai printar todos os nomes por estar percorrendo o dicionario
for name in favorite_languages:
  print(f"Olá {name}")

  #o caso do if é se name vai estar em friends faça uma variavel
  if name in friends:
    #essa variavel vai entregar uma mensagem especial para que passou no if, a variavel vai chamar os nomes do dicionario que são a chave

    language = favorite_languages[name].title() #name é a variavel do for
    print(f"{name} ama essa linguagem {language}")
    #o print inclui as variaveis name e language
```

```python
#Criando uma frota de 30 alienigina usando o RANGE()

#Essa vai ser a lista aonde vai armazenar a chave e valor do dicionario
aliens = []

#a função do range é retornar a serie de numeros informando quantas vezes deve repitir
for alien_number in range(0,30):

  #essa variavel serve para armazenar a chave e o valor do dicionario da horda de aliens e o valor retornado do range
  new_alien = {'color': 'green',
               'speed': 'faster',
               'points': 10
               }

  #serve para adicionar a variavel na lista aliens, ainda esta no for
  aliens.append(new_alien)

#Aqui usamos um laço for para mostrar a quantidade total dos aliens, mas como usamos um fatiador apenas mostrara 5 aliens
for alien in aliens[0:5]:
  print(alien)

# aqui usamos o len para medir quantos aliens foi criado
print(f'o total de aliens foi {len(aliens)}')
```

```python
# 1. Nosso jogador representado por um Dicionário
player = {
    'nome': 'pablosempai',
    'idade': 50,
    'nivel': 17,
    'maca': 1  # Quantidade de maçãs no inventário
}

print(f"=== BEM-VINDO AO RPG, {player['nome'].upper()} ===")
print(f"Seu nível atual é: {player['nivel']}")

# 2. Verificação com IF / ELSE
if player['nivel'] < 18:
    print("\n[AVISO] Você precisa estar no nível 18 para entrar na caverna do dragão!")
    print(f"Você tem {player['maca']} maçã(s) no inventário.")

    # 3. Decisão do jogador usando INPUT
    decisao = input("Deseja comer uma maçã para subir de nível? (sim/nao): ").lower()

    # 4. Processando a escolha com IF-ELIF-ELSE
    if decisao == 'sim':
        if player['maca'] > 0:
            player['maca'] -= 1        # Gasta 1 maçã
            player['nivel'] += 1       # Aumenta 1 nível (igual a: player['nivel'] = player['nivel'] + 1)

            print("\nVocê comeu a maçã dourada! 🍎")
            print(f"PARABÉNS! Seu nível subiu para: {player['nivel']}")
            print("Entrada da caverna LIBERADA!")
        else:
            print("\nVocê não tem maçãs no inventário!")

    elif decisao == 'nao':
        print("\nVocê decidiu não comer. Continue treinando no nível 17!")

    else:
        print("\nOpção inválida! Digite apenas 'sim' ou 'nao'.")

else:
    print("\nVocê já é nível 18 ou maior! A caverna está liberada!")
```
